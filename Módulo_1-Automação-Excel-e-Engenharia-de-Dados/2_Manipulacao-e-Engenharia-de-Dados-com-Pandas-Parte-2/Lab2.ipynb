## Processamento de Dados com Método Apply

O método apply() no contexto de bibliotecas de manipulação de dados como pandas em Python é extremamente útil para aplicar uma função ao longo de um eixo de um DataFrame ou em elementos de uma Series. Ele permite uma flexibilidade significativa, sendo capaz de operar tanto em uma linha ou coluna inteira de um DataFrame, quanto em cada elemento individualmente de uma Series.

```python
# Slicing do dataframe
df_dsa2[df_dsa2['input2'] < 50].iloc[1:5:,1:4]
```

```python
# Aplica a função sum no eixo 0, que corresponde ao eixo das linhas. 
# Isso significa que a soma será calculada verticalmente para cada coluna no intervalo selecionado. 
# O resultado será uma série contendo a soma de cada uma dessas colunas.
df_dsa2[df_dsa2['input2'] < 50].iloc[1:5:,1:4].apply(sum, axis = 0)
```

```python
# Mesmo resultado anterior
df_dsa2[df_dsa2['input2'] < 50].iloc[1:5:,1:4].apply(sum)
```

```python
# Aqui, a função sum é aplicada no eixo 1, que corresponde ao eixo das colunas. 
# Isso significa que a soma será calculada horizontalmente para cada linha no intervalo de colunas selecionado. 
# O resultado será uma série contendo a soma dos valores em cada linha para as colunas selecionadas.
df_dsa2[df_dsa2['input2'] < 50].iloc[1:5:,1:4].apply(sum, axis = 1)
```

### Método Apply com Expressão Lambda 

Expressão Lambda é função anônima em Python (existe somente em tempo de execução).

As expressões lambda em Python são uma maneira concisa de declarar funções anônimas, ou seja, funções sem um nome definido. Elas são particularmente úteis quando você precisa de uma função simples por um curto período de tempo e não quer a formalidade de definir uma função com def. As expressões lambda são muitas vezes usadas em conjunto com funções de ordem superior, como map(), filter() e reduce(), bem como métodos como apply() em pandas.

```python
# Criando um DataFrame com duas colunas de números
df = pd.DataFrame({
    'A': [10, 20, 30],
    'B': [40, 50, 60]
})
```

```python
# Aplicando a função lambda para dobrar cada valor
doubled_df = df.apply(lambda x: x * 2)
```

```python
print(doubled_df)
```

```python
# Cria uma função
def dsa_converte_dados(x):
    if x < 10:
        return 'G10'
    elif x < 50:
        return 'G50'
    elif x < 90:
        return 'G90'
    else:
        return 'NO'
```

```python
df_dsa2['input4']
```

```python
df_dsa2['input4'].apply(lambda x: dsa_converte_dados(x))
```

```python
df_dsa2['input4'].apply(dsa_converte_dados)
```

```python
df_dsa2['input4'].map(dsa_converte_dados)
```

```python
# Tempo de execução
a = datetime.datetime.now()
df_dsa2['input4'].apply(lambda x: dsa_converte_dados(x))
b = datetime.datetime.now()
print (b-a)
```

```python
# Tempo de execução
a = datetime.datetime.now()
df_dsa2['input4'].apply(dsa_converte_dados)
b = datetime.datetime.now()
print (b-a)
```

```python
# Tempo de execução
a = datetime.datetime.now()
df_dsa2['input4'].map(dsa_converte_dados)
b = datetime.datetime.now()
print (b-a)
```

Para a maioria dos casos onde a função aplicada é simples e direta, map() é uma escolha melhor em termos de velocidade.

```python
df_dsa2.map(dsa_converte_dados).head()
```

## Dicing de Dataframes

Para exemplificar o "dicing" com pandas, vamos criar um pequeno DataFrame e então selecionar dados específicos dele, combinando condições em múltiplas colunas. "Dicing" é um termo usado para descrever a seleção de uma subseção específica de um conjunto de dados, baseando-se em critérios específicos, similar ao que faríamos ao cortar um dado (ou "dice") em partes menores.

Primeiro, criaremos um DataFrame pandas com dados fictícios:

```python
# Criando um DataFrame de exemplo
dados_dsa = {
    'Nome': ['Ana', 'Bruno', 'Carla', 'Diego', 'Eva', 'Fábio'],
    'Idade': [23, 34, 45, 26, 29, 35],
    'Cidade': ['São Paulo', 'Rio de Janeiro', 'São Paulo', 'Curitiba', 'Rio de Janeiro', 'Curitiba']
}
```

```python
# Converte o dicionário em dataframe
df = pd.DataFrame(dados_dsa)
```

```python
# Mostra o DataFrame criado
print(df)
```

```python
# Selecionando pessoas com mais de 30 anos que vivem no Rio de Janeiro
filtro = (df['Idade'] > 30) & (df['Cidade'] == 'Rio de Janeiro')
```

```python
df_filtrado = df[filtro]
```

```python
# Mostrando o resultado do dicing
print(df_filtrado)
```

## Merge, Join e Concatenação

```python
# DataFrames de exemplo para Merge e Join

df_a = pd.DataFrame({
    'id': [1, 2, 3, 4],
    'Nome': ['Ana', 'Bruno', 'Carla', 'Diego']
})

df_b = pd.DataFrame({
    'id': [2, 3, 4, 5],
    'Idade': [34, 45, 26, 29]
})

# DataFrame adicional para Concatenação
df_c = pd.DataFrame({
    'id': [6, 7],
    'Nome': ['Eva', 'Fábio'],
    'Idade': [32, 28]
})
```

```python
# Merge (Inner Join) baseado na coluna 'id'
df_merged = pd.merge(df_a, df_b, on = 'id')
```

```python
print("Merge (Inner Join):\n", df_merged)
```

```python
# Definindo 'id' como índice para usar no Join
df_a_indexed = df_a.set_index('id')
df_b_indexed = df_b.set_index('id')
```

```python
# Join dos DataFrames no índice
df_joined = df_a_indexed.join(df_b_indexed, how = 'inner')
```

```python
print("Join baseado em índice:\n", df_joined)
```

```python
# Join dos DataFrames no índice
df_joined = df_a_indexed.join(df_b_indexed, how = 'outer')
```

```python
print("Join baseado em índice:\n", df_joined)
```

```python
# Concatenação vertical (adicionando linhas)
df_concat_v = pd.concat([df_a, df_c], ignore_index = True)
```

```python
print("Concatenação Vertical:\n", df_concat_v)
```

> Para concatenação horizontal, assegure-se de que os DataFrames compartilham o mesmo índice ou redefina-os conforme necessário.
